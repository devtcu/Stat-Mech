```python
from random import random, randrange
from math import exp, pi
from numpy import ones
import matplotlib.pyplot as plt
```

```python
T = 10
N = 1000
steps = 250000
n = ones([N,3],int)
eplot = []
E = 3*N*pi*pi/2
for k in range(steps):
    i = randrange(N)
    j = randrange(3)
    if random()<0.5:
        dn = 1
        dE = (2*n[i,j]+1)*pi*pi/2
    else:
        dn = -1
        dE = (-2*n[i,j]+1)*pi*pi/2
    if n[i,j]>1 or dn == 1:
        if random()<exp(-dE/T):
            n[i,j] += dn
            E += dE
    eplot.append(E)
```

```python
plt.plot(eplot)
plt.title("Energy vs Step")
plt.ylabel("Energy")
plt.show()
```
